```python
import pandas as pd

df = pd.read_csv("2001.csv", low_memory=False)

# Inspect the actual column names first
print(df.columns.tolist())
```

```python
# Filter records for one commodity
onion_data = df[
    df["Commodity"].astype("string").str.strip().str.lower()
    == "onion"
]

# Select records with a positive modal price
valid_price_data = df[
    pd.to_numeric(df["Modal_Price"], errors="coerce") > 0
]

print("Onion records:", len(onion_data))
print("Positive-price records:", len(valid_price_data))
```

```python
###Normalization
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df[["Modal_Price_Normalized"]] = scaler.fit_transform(
    df[["Modal_Price"]]
)
```

```python
### Feature engineering
```

```python
df["Arrival_Date"] = pd.to_datetime(
    df["Arrival_Date"], errors="coerce"
)

df["Year"] = df["Arrival_Date"].dt.year
df["Month"] = df["Arrival_Date"].dt.month
df["Day_of_Week"] = df["Arrival_Date"].dt.dayofweek
df["Quarter"] = df["Arrival_Date"].dt.quarter
```

```python
df["Price_Range"] = (
    df["Max_Price"] - df["Min_Price"]
)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
###removing the missing values(Max_Price-3,Price_Range-3)
df['Max_Price'] = df['Max_Price'].fillna(df['Max_Price'].mean())
df['Price_Range'] = df['Price_Range'].fillna(df['Price_Range'].mean())
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
###outliers 
col = "Modal_Price"
q1 = df[col].quantile(0.25)
q3 = df[col].quantile(0.75)
iqr = q3 - q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

df["Modal_Price_outlier_flag"] = (
    (df[col] < lower) | (df[col] > upper)
)

print(df["Modal_Price_outlier_flag"].value_counts())
```
